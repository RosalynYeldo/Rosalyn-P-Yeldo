```r
# PHASE 1 - EXPLORATORY DATA ANALYSIS
data <- read.csv("https://raw.githubusercontent.com/salemprakash/EDA/main/Data/NaturalGas.csv")

# View first and last few records
head(data)
tail(data)

# Dataset dimensions
dim(data)

# Column names
colnames(data)

# Structure of dataset
str(data)
```

```r
summary(data)

# Mean of numeric columns
sapply(data, function(x)
  if(is.numeric(x)) mean(x, na.rm=TRUE))

# Median
sapply(data, function(x)
  if(is.numeric(x)) median(x, na.rm=TRUE))

# Standard Deviation
sapply(data, function(x)
  if(is.numeric(x)) sd(x, na.rm=TRUE))

# Variance
sapply(data, function(x)
  if(is.numeric(x)) var(x, na.rm=TRUE))

# Minimum
sapply(data, function(x)
  if(is.numeric(x)) min(x, na.rm=TRUE))

# Maximum
sapply(data, function(x)
  if(is.numeric(x)) max(x, na.rm=TRUE))

# Quartiles
sapply(data, function(x)
  if(is.numeric(x)) quantile(x, na.rm=TRUE))
```

```r
# Count missing values in each column
colSums(is.na(data))

# Total missing values
sum(is.na(data))

# Display rows containing missing values
data[!complete.cases(data), ]


# Method 1 : Remove missing values
clean_data <- na.omit(data)


# Method 2 : Forward Fill
forward_fill <- data.frame(
  lapply(data, function(x)
    na.locf(x, na.rm = FALSE))
)


# Method 3 : Backward Fill
backward_fill <- data.frame(
  lapply(data, function(x)
    na.locf(x,
            fromLast = TRUE,
            na.rm = FALSE))
)


# Method 4 : Mean Imputation
mean_impute <- data

numeric_columns <- sapply(mean_impute, is.numeric)

mean_impute[numeric_columns] <-
  lapply(mean_impute[numeric_columns], function(x){

    x[is.na(x)] <- mean(x, na.rm = TRUE)
    return(x)

  })
```

```r
# 5. DATA CLEANING


# Remove duplicate rows
clean_data <- distinct(clean_data)

# Check duplicates
sum(duplicated(clean_data))

# Check data types
str(clean_data)

# Convert Date column if present
if("Date" %in% names(clean_data))
{
  clean_data$Date <- as.Date(clean_data$Date)
}
```

```r
# 6. DATA TRANSFORMATION


# Select numeric columns
numeric_data <- clean_data %>%
  select(where(is.numeric))

# Normalization (Min-Max Scaling)
normalized_data <- as.data.frame(
  lapply(numeric_data, function(x){

    (x-min(x))/(max(x)-min(x))

  })
)

# Standardization (Z-score)
standardized_data <- as.data.frame(
  scale(numeric_data)
)
```

```r
# 7. UNIVARIATE ANALYSIS

# Identify first numeric column
num_col <- names(numeric_data)[1]


# Visualization 1 : Histogram
ggplot(clean_data,
       aes(x=.data[[num_col]]))+

  geom_histogram(
    bins=20,
    fill="steelblue",
    color="black")+

  ggtitle("Histogram")


# Visualization 2 : Box Plot
ggplot(clean_data,
       aes(y=.data[[num_col]]))+

  geom_boxplot(fill="orange")+

  ggtitle("Box Plot")


# Visualization 3 : Density Plot
ggplot(clean_data,
       aes(x=.data[[num_col]]))+

  geom_density(fill="lightgreen",
               alpha=0.6)+

  ggtitle("Density Plot")
```

```r
# 8. BIVARIATE ANALYSIS (Minimum 3 Visualizations)


# Select first two numeric columns
x <- names(numeric_data)[1]
y <- names(numeric_data)[2]


# Visualization 1 : Scatter Plot
ggplot(clean_data,
       aes(x=.data[[x]],
           y=.data[[y]]))+

  geom_point(color="blue")+

  ggtitle("Scatter Plot")


# Visualization 2 : Regression Plot
ggplot(clean_data,
       aes(x=.data[[x]],
           y=.data[[y]]))+

  geom_point()+

  geom_smooth(method="lm",
              se=FALSE,
              color="red")+

  ggtitle("Scatter Plot with Regression Line")


# Visualization 3 : Grouped Boxplot
ggplot(clean_data,
       aes(x=factor(cut(.data[[x]],4)),
           y=.data[[y]]))+

  geom_boxplot(fill="skyblue")+

  xlab(x)+

  ggtitle("Grouped Boxplot")
```

```r
# 9. MULTIVARIATE ANALYSIS (Minimum 3 Visualizations)

# Visualization 1 : Correlation Heatmap

correlation_matrix <- cor(numeric_data)

corrplot(correlation_matrix,
         method="color",
         tl.cex=0.8)



# Visualization 2 : Pair Plot

ggpairs(numeric_data)



# Visualization 3 : Scatter Plot Matrix using Base R

pairs(numeric_data,
      main="Scatter Plot Matrix",
      pch=19,
      col="darkgreen")
```

```r
# Install packages if required
# install.packages("e1071", repos = "https://cloud.r-project.org")
# 1D, 2D, and 3D statistical analyses

library(e1071)

#1
#data <- read.csv("https://raw.githubusercontent.com/salemprakash/EDA/main/Data/NaturalGas.csv")

#2
mean(data$consumption)
median(data$consumption)
min(data$consumption)
max(data$consumption)
range(data$consumption)
var(data$consumption)
sd(data$consumption)
IQR(data$consumption)
skewness(data$consumption)
kurtosis(data$consumption)

#3
stats <- c(
  mean(data$consumption),
  median(data$consumption),
  min(data$consumption),
  max(data$consumption),
  range(data$consumption),
  var(data$consumption),
  sd(data$consumption),
  IQR(data$consumption),
  skewness(data$consumption),
  kurtosis(data$consumption)
)

stats

#4
freq <- table(data$state)
prop.table(freq)

pie(freq,
    main = "Pie Chart of Frequency Distribution of State")

barplot(freq,
        main = "Bar Chart of Frequency Distribution of State",
        xlab = "States",
        ylab = "Freq")

#5
fnum <- cut(data$consumption, breaks = 10)
table(fnum)

hist(data$consumption,
     breaks = 10,
     main = "Histogram of Consumption",
     xlab = "Consumption")

barplot(data$consumption,
        main = "Bar Plot of Consumption",
        ylab = "Consumption")

#6
data$statecode <- as.factor(data$statecode)

cross <- table(data$state, data$statecode)

prop.table(cross, margin = 1)
prop.table(cross, margin = 2)
prop.table(cross)

#7
barplot(cross,
        beside = FALSE,
        main = "State and State Code Distribution",
        xlab = "State",
        ylab = "Freq")

#8
aggregate(consumption ~ state,
          data = data,
          FUN = mean)

aggregate(consumption ~ state,
          data = data,
          FUN = median)

aggregate(consumption ~ state,
          data = data,
          FUN = sd)

#9
boxplot(consumption ~ state,
        data = data,
        main = "Boxplot of Consumption Across Different States")

library(ggplot2)

ggplot(data, aes(x = state, y = consumption)) +
  geom_violin() +
  labs(title = "Violin Plot of Consumption Across Various States",
       x = "States",
       y = "Consumption") +
  theme(axis.text.x = element_text(angle = 90))

#10
cor(data$consumption,
    data$price,
    method = "pearson")

cor(data$consumption,
    data$price,
    method = "spearman")

plot(data$price,
     data$consumption,
     xlab = "Price",
     ylab = "Consumption",
     main = "Price VS Consumption")

abline(lm(consumption ~ price, data = data))

#11
library(dplyr)

summary_data <- data %>%
  group_by(state, statecode) %>%
  summarise(
    mean = mean(consumption),
    sd = sd(consumption),
    .groups = "drop"
  )

ggplot(summary_data,
       aes(x = state, y = mean, fill = statecode)) +
  geom_col(position = "dodge") +
  geom_errorbar(
    aes(ymin = mean - sd,
        ymax = mean + sd),
    position = position_dodge(0.9),
    width = 0.2
  ) +
  theme(axis.text.x = element_text(angle = 90))

#12
ggplot(data,
       aes(x = price,
           y = consumption,
           colour = state,
           shape = statecode)) +
  geom_point() +
  labs(title = "Scatter Plot of Consumption to Price Across Different States and Statecodes")

cor_m <- aggregate(
  data[, c("price", "consumption")],
  by = list(state = data$state),
  FUN = function(x) x
)

cor_m

#13
ndata <- data[, c("price", "consumption", "heating")]

cm <- cor(ndata)

heatmap(cm,
        main = "Heatmap of Price, Consumption and Heating")

#14
pairs(data[, c("consumption", "price", "heating")],
      main = "Pair Plot")
```

```r
# K-Means Clustering

#data <- read.csv("https://raw.githubusercontent.com/salemprakash/EDA/main/Data/NaturalGas.csv")

# Select numerical variables
cdata <- data[, c("consumption", "price", "eprice",
                  "oprice", "lprice", "heating", "income")]

# Standardize the data
sdata <- scale(cdata)

# Apply K-Means clustering
km <- kmeans(sdata, centers = 3)

# Display cluster assignments
print(km$cluster)

# Display cluster sizes
print(table(km$cluster))

# Display cluster centers
print(km$centers)

```

```r
# Hierarchical Clustering

#data <- read.csv("https://raw.githubusercontent.com/salemprakash/EDA/main/Data/NaturalGas.csv")

# Select numerical variables
cdata <- data[, c("consumption", "price", "eprice",
                  "oprice", "lprice", "heating", "income")]

# Standardize the data
sdata <- scale(cdata)

# Calculate Euclidean distance
d <- dist(sdata, method = "euclidean")

# Apply Hierarchical Clustering
hc <- hclust(d, method = "complete")

# Plot dendrogram
plot(hc, main = "Hierarchical Clustering")

# Extract 3 clusters
cluster <- cutree(hc, k = 3)

# Display cluster assignments
print(cluster)

# Display cluster sizes
print(table(cluster))
```
